```python
import pandas as pd
import hashlib

data = pd.read_csv("C:/Users/SAI NITISH/OneDrive/Desktop/bosch_project2/data/product_quality_data.csv")
data.head()
```

```python
data.tail()
```

```python
def check_product_authenticity(row):
    if row["Packaging_Integrity"] == "No" or row["Serialization_Status"] == "No":
        return "Counterfeit"
    elif row["Inspection_Score"] < 75:
        return "Genuine - High Risk"
    else:
        return "Genuine - Low Risk"
```

```python
def generate_blockchain_hash(row, authenticity_status):

    record = (
        row["Product_ID"]
        + row["Batch_ID"]
        + row["Component_Type"]
        + row["Supplier_ID"]
        + str(row["Inspection_Score"])
        + row["Packaging_Integrity"]
        + row["Serialization_Status"]
        + authenticity_status
    )

    return hashlib.sha256(record.encode()).hexdigest()
```

```python
product_id = "P302-04"   # change this to test

row = data[data["Product_ID"] == product_id]

if row.empty:
    print("❌ Product not found")
else:
    row = row.iloc[0]

    status = check_product_authenticity(row)
    blockchain_hash = generate_blockchain_hash(row, status)

    print("Product ID:", row["Product_ID"])
    print("Batch ID:", row["Batch_ID"])
    print("Component Type:", row["Component_Type"])
    print("Supplier ID:", row["Supplier_ID"])
    print("Inspection Score:", row["Inspection_Score"])
    print("Packaging Integrity:", row["Packaging_Integrity"])
    print("Serialization Status:", row["Serialization_Status"])
    print("Authenticity Status:", status)
    print("Blockchain Hash:", blockchain_hash)
```

```python
for _, row in data.head(5).iterrows():
    status = check_product_authenticity(row)
    hash_value = generate_blockchain_hash(row, status)

    print(row["Product_ID"], "→", status)
    print("Hash:", hash_value)
    print("-" * 50)
```

```python
for _, row in data.tail(5).iterrows():
    status = check_product_authenticity(row)
    hash_value = generate_blockchain_hash(row, status)

    print(row["Product_ID"], "→", status)
    print("Hash:", hash_value)
    print("-" * 50)
```
